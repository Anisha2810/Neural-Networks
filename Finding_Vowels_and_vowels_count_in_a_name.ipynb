```python
# Program to find vowels in a name and count them

name = input("Enter your name: ")

vowels = "aeiouAEIOU"
count = 0

print("Vowels in the name are:")

for letter in name:
    if letter in vowels:
        print(letter)
        count += 1

print("Total number of vowels:", count)
```

```python
# Alternative program to find vowels in a name and count them

name = input("Enter your name: ")

vowels = "aeiouAEIOU"

# Find all vowels in the name using list comprehension
found_vowels = [letter for letter in name if letter in vowels]

print("Vowels in the name are:", ", ".join(found_vowels))
print("Total number of vowels:", len(found_vowels))
```
