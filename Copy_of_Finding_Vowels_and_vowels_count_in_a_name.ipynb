```python
# Program to find vowels in a name and count them

name = input("Enter your name: ")

vowels = "aeiouAEIOU"
count = 0

print("Vowels in the name are:")

for letter in name:
    if letter in vowels:
        print(letter)
        count += 1

print("Total number of vowels:", count)
```

```python
# Alternative program to find vowels in a name and count them

name = input("Enter your name: ")

vowels = "aeiouAEIOU"

# Find all vowels in the name using list comprehension
found_vowels = [letter for letter in name if letter in vowels]

print("Vowels in the name are:", ", ".join(found_vowels))
print("Total number of vowels:", len(found_vowels))
```

```python
# Program to find vowels in a name using a dictionary

name = input("Enter your name: ")

vowels = "aeiou"
vowel_count = {}  # empty dictionary to store counts

for letter in name.lower():  # convert all letters to lowercase
    if letter in vowels:
        if letter in vowel_count:
            vowel_count[letter] += 1
        else:
            vowel_count[letter] = 1

print("Vowels in the name with their counts:")
for vowel, count in vowel_count.items():
    print(f"{vowel}: {count}")

total_vowels = sum(vowel_count.values())
print("Total number of vowels:", total_vowels)
```
